```python
from ipywidgets import interact, FloatSlider
import matplotlib.pyplot as plt
import ipywidgets as widgets
import sys
sys.path.insert(1,'../scripts')
sys.path.insert(1,'../stella')
from models import comet_curve2
from utils import *
import stella
```

```python

def plot_transit(sigma, tail, shape):
    plt.figure(figsize=(8, 4))
    model = 1 - comet_curve2(time, amplitude, t0, sigma=sigma, tail=tail, shape=shape)
    plt.plot(time, model)
    #plt.ylim(0.95, 1.01)
    plt.title(f'σ={sigma:.2f}, tail={tail:.2f}, shape={shape:.2f}')
    plt.show()
```

```python
time = np.linspace(1470,1490,1200) 
t0 = 1475
amplitude = 1


sigma = np.random.uniform(0.25, 0.7, size=1000) 
tail = np.random.uniform(0.35, 1, size=1000)
shape = np.random.uniform(1.5, 4.0, size=1000)


interact(plot_transit,
        sigma=FloatSlider(min=0.25, max=1.5, step=0.05, value=0.5),
        tail=FloatSlider(min=0.35, max=1, step=0.05, value=0.5),
        shape=FloatSlider(min=1.0, max=10.0, step=0.2, value=2.5))
```

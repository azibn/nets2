```python

def plot_transit(sigma, tail, shape):
    plt.figure(figsize=(8, 4))
    model = 1 - comet_curve2(time, amplitude, t0, sigma=sigma, tail=tail, shape=shape)
    plt.plot(time, model)
    #plt.ylim(0.95, 1.01)
    plt.title(f'σ={sigma:.2f}, tail={tail:.2f}, shape={shape:.2f}')
    plt.show()
```

```python
def ldecomet(t, K, beta, t0, delta_t):
    # Calculate Δ
    delta = np.where(t >= t0, beta * (t - t0), 0)
    
    # Calculate Δ'
    delta_prime = np.where(t >= t0 + delta_t, beta * (t - t0 - delta_t), 0)
    
    # Calculate the relative flux decrease
    delta_f_over_f = K * (np.exp(-delta) - np.exp(-delta_prime))
    
    return delta_f_over_f
```
